```python
import numpy as np
a = np.random.default_rng().standard_normal(1<<30, 'f')
a.cumsum(out=a)
a /= np.sqrt(*[np.linspace(2/np.pi, 2/np.pi*len(a), len(a), dtype='f')]*2)
a -= a.min()
a /= a.max()
a = np.diff(a)
```

```python
def frexp(a):
    if a.dtype == 'f':
        a = a.view('B')[3::4]&128, (a.view('H')[1::2]<<1).view('B')[1::2], a.view('I')&8388607
    else:
        a = a.view('B')[1::2]&128, a.view('B')[1::2]&124, a.view('H')&1023
    return a
def entropy(a):
    m = np.zeros(a.max()+1, 'I')
    np.add.at(m, a, 1)
    m = m[m.astype('?')]
    m = np.divide(m, m.sum(), m.view('f'))
    return np.log2(m) @ m
```

```python
list(map(entropy, frexp(a.astype('e'))))
```

```python
def frexp(f, a, s=0):
    f = open(f, 'wb')
    (a.view('I')&2**24-1)[:,None].view('B')[:,:3].tofile(f)
    (a.view('H')[1::2]<<1).view('B')[1::2].tofile(f)
    if s:
        (a.view('B')[3::4]&2**7).tofile(f)
frexp('Untitled3.raw', a)
frexp('Untitled3.diff', np.diff(a, prepend=0).cumsum().astype('f'))
```

```python
def save_lfoat(f, a):
    f = open(f, 'wb')
    d,c,b,a = a[:,None].view('B').T
    t = a & 127
    t |= b & 128
    t.tofile(f)
    t = b & 127
    t |= a & 128
    t.tofile(f)
    c.tofile(f)
    d.tofile(f)
```

```python
save_lfoat('Untitled3.raw', a)
save_lfoat('Untitled3.diff', np.diff(a))
```
