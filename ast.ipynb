```python
from ast import *
def check(src, name=['A电场电压'], op=(Add,Sub,Mult,Div)):
    def check(a):
        if isinstance(a, BoolOp):
            for a in a.values:
                check(a)
        elif isinstance(a, Compare):
            if len(a.ops) > 1:
                raise SyntaxError(get_source_segment(src, a))
            check(a.left)
            check(a.comparators[0])
        elif isinstance(a, BinOp):
            if not isinstance(a.op, op):
                raise SyntaxError(get_source_segment(src, a))
            check(a.left)
            check(a.right)
        elif isinstance(a, Call):
            if a.func.id not in fn:
                raise NameError(a.func.id)
            if len(a.args) != 1:
                raise SyntaxError(get_source_segment(src, a))
            check(a.args[0])
        elif isinstance(a, Name):
            if a.id not in name:
                raise NameError(a.id)
        elif isinstance(a, Constant):
            if not isinstance(a.n, (float,int)):
                raise SyntaxError(get_source_segment(src, a))
        else:
            raise SyntaxError(get_source_segment(src, a))
    if len(src) > 99:
        raise Exception('too long')
    a = parse(src).body
    if len(a) != 1 or not isinstance(a[0], Expr):
        raise SyntaxError(src)
    check(a[0].value)
    return eval(src, {**name,**{i.__name__:i for i in fn}})
```
