```python
src = """
def f():
    return a.b(3)
"""
```

```python
from ast import *
```

```python
a = parse(src)
```

```python
a.body[0].body[0].value.func = wait(a.body[0].body[0].value.func)
a.body[0].body[0].value = wait(a.body[0].body[0].value)
```

```python
print(unparse(a))
```
