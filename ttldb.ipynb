```python
from asyncio import get_event_loop
l = get_event_loop()
class Db(dict):
    __slots__ = ()
    def insert(s, i, v, t):
        s[i] = v
        l.call_later(t, s.__delitem__, i)
```

```python
db = Db()
```

```python
from random import random
for i in  range(99999999):
    db.insert(i, i, random()*100)
```
