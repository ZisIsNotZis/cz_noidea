```python
from numba import njit, guvectorize
import numpy as np
@njit('f4[:,:](f4[:,:],f4[:,:])')
def gemm(a, b):
    c = np.zeros((len(a),len(b)), np.float32)
    for i in range(len(a)):
        for j in range(len(b)):
            c[i,j] = a[i,0] * b[j,0]
            for k in range(1, a.shape[1]):
                c[i,j] += a[i,k] * b[j,k]
    return c
@guvectorize('void(f4[:],f4[:],f4[:])', '(x),(x)->()', nopython=1)
def gumm(a, b, c):
    t = a[0] * b[0]
    for i in range(1, len(a)):
        t += a[i] * b[i]
    c[0] = t
```

```python
%timeit gumm(a, b[:,None])
```
