```python
from pynng import Push0, Pull0
from asyncio import gather
from time import time
```

```python
addr = 'ipc:///tmp/nng.ipc'
async def push():
    s = Push0(dial=addr)
    t = []
    b = bytes(67108864)
    for _ in range(9):
        for _ in range(9):
            await s.asend(b)
        t.append(time())
    return t
async def pull():
    t = []
    for _ in range(9):
        for _ in range(9):
            await s.arecv()
        t.append(time())
    return t
with Pull0(listen=addr) as s:
    t = await gather(push(), pull())
```

```python
from zmq import PUB, SUB
from zmq.asyncio import Context
addr = 'ipc:///tmp/zmq.ipc'
c = Context()
async def push():
    s = c.socket(PUB)
    s.connect(addr)
    t = []
    b = bytes(67108864)
    for _ in range(9):
        for _ in range(9):
            await s.send_multipart(b)
        t.append(time())
    return t
async def pull():
    t = []
    for _ in range(9):
        for _ in range(9):
            await s.recv_multipart()
        t.append(time())
    return t
with c.socket(SUB) as s:
    s.bind(addr)
    t = await gather(push(), pull())
```
