```python
from inspect import getsource
from ast import parse
from random import shuffle
def 乱序执行(f):
    f = parse(getsource(f).partition('\n')[2], mode='single')
    def 乱序执行(*a, **k):
        shuffle(f.body[0].body)
        t = {}
        eval(compile(f, '', 'single'), t)
        return t[f.body[0].name](*a, **k)
    return 乱序执行
```

```python
@乱序执行
def f():
    print('hello')
    print('world')
    print('!')
    for _ in range(3):
        print('for')
f()
```

```python
from asyncio import get_event_loop
def 延迟执行(f):
    def 延迟执行(*a):
        get_event_loop().call_soon(f, *a, **k)
    return 延迟执行
```

```python
@延迟执行
def f(a):
    print(a)
print('=== begin ===')
f(1)
print(2)
f(3)
print(4)
print('=== end ===')
```

```python
from asyncio import get_event_loop
from datetime import datetime
def 延迟执行(t):
    def 延迟执行(f):
        c = None
        def 延迟执行(*a):
            nonlocal c
            if c:
                c.cancel()
            c = get_event_loop().call_later(t, f, *a)
        return 延迟执行
    return 延迟执行
def 现在几点(f):
    def 现在几点(*a):
        print(f'{f.__name__}({", ".join(map(str,a))})', '开始:', datetime.now())
        f(*a)
        print(f'{f.__name__}({", ".join(map(str,a))})', '结束:', datetime.now())
    return 现在几点
```

```python
@现在几点
@延迟执行(5)
@现在几点
def f(a, b, c):
    print(a)
    print(b)
    print(c)
f(3)
```
