```python
from threading import Thread
def Go(f):
    Thread(target=f).start()
```

```python
from asyncio import get_event_loop, new_event_loop, run_coroutine_threadsafe, wrap_future, wait
from os import cpu_count
from random import choice
class go(list):
    def __init__(s):
        s.p = [get_event_loop()]
        for _ in range(cpu_count()-1):
            @Go
            def _():
                l = new_event_loop()
                s.p.append(l)
                l.run_forever()
    def __call__(s, *a):
        def _(f):
            s.append(wrap_future(run_coroutine_threadsafe(f(*a), choice(s.p))))
        return _
    async def ne(s):
        r = []
        while s:
            t = tuple(s)
            s.clear()
            r.extend([await t for t in (await wait(t))[0]])
        return r
go = go()
```

```python
import numpy as np
for _ in range(64):
    @go()
    async def f():
        t = np.random.rand(4096,4096).astype('f')
        t = t @ t
        t = t @ t
        return (t @ t).mean()
np.mean(await go.ne())
```
