```python
import numpy as np
time = np.random.rand(12,4,10)
max = time + np.random.rand()/10
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
def compute_big_fig():
    fig = make_subplots()
    fig.add_trace(go.Bar(y=time.sum((1,2))))
    fig.add_trace(go.Scatter(y=max.sum((1,2))))
    return fig
def compute_small_fig(n):
    fig = make_subplots()
    fig.add_trace(go.Bar(y=time[n].sum(1)))
    fig.add_trace(go.Scatter(y=max[n].sum(1)))
    return fig
```

```python
from dash import Dash
from dash.dependencies import Input, Output, State
from dash_core_components import Graph
from dash_html_components import Div, H3
from dash.exceptions import PreventUpdate

app = Dash()

app.layout = Div([
    H3(id='title', children='workhour / month'),
    Graph(id='graph', figure=compute_big_fig())
])

@app.callback([Output('graph', 'figure'), Output('title', 'children')],
              [Input('graph', 'clickData')],
              [State('title', 'children')])

def change_fig(graph, title):
    if title == 'workhour / month':
        graph = graph['points'][0]['x']
        return compute_small_fig(graph), f'workhour / team @ {graph}th month'
    return compute_big_fig(), 'workhour / month'

app.run_server(host='0.0.0.0')
```
