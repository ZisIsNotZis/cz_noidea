```python
import numpy as np
a = np.random.default_rng().standard_normal(dtype='f', out=np.memmap('/datb/a.f', 'f', 'w+', shape=2**32))
```

```python
from numba import njit
@njit('void(f4[:])')
def decsum(a):
    for i in range(1, len(a)):
        a[i] = a[i-1] * .9999999 + a[i]
decsum(a)
```

```python
import matplotlib.pyplot as plt
plt.plot(a[::1000000])
```

```python
a -= a.min()
a *= 255.999 / a.max()
A = np.memmap('/datb/raw.B', 'B', 'w+', shape=len(a))
A[:] = a
B = np.memmap('/datb/diff.B', 'B', 'w+', shape=len(a)-1)
np.subtract(A[1:], A[:-1], B)
!rm /datb/a.f
del a
```

```python
def entropy(a):
    t = np.zeros(256, 'I')
    np.add.at(t, a, 1)
    t = t / len(a)
    return t @ np.log(t + 1e-9) / 8
entropy(A), entropy(B)
```
