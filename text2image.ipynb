```python
from PIL import Image, ImageDraw, ImageFont, ImageChops
from flask import Flask, send_file

app = Flask(__name__)

@app.route('/<int:w>/<int:h>/<a>')
def f(w, h, a):
    i = Image.new('RGB', (64*len(a),64))
    ImageDraw.Draw(i).text((0,0), a, font=ImageFont.truetype('/usr/share/fonts/truetype/wqy/wqy-microhei.ttc', 56))
    ImageChops.invert(i.crop(i.getbbox()).resize((w,h))).save('/tmp/text2image.jpg')
    return send_file('/tmp/text2image.jpg', mimetype='image/jpeg')

app.run('0.0.0.0', '11111')
```
