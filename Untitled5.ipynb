```python
from operator import *
from asyncio import Future, create_task
from inspect import isawaitable
def call(f, *a):
    return f(*a)
def go(f):
    create_task(f())
async def op(f, *a):
    return f(*[await a if isawaitable(a) else a for a in a])
class F:
    def __init__(s, a=0):
        s._a = create_task(a) if a else Future()
    def __await__(s):
        return s._a.__await__()
    def set(s, a):
        s._a.set_result(a)
    for _ in add,sub,mul,truediv,floordiv,divmod,call,getitem,setitem,setattr:
        locals()[_.__name__] = lambda *a: op(f, *a)
    for _ in add,sub,mul,truediv,floordiv,divmod,call,getitem,setitem,setattr:
        locals()[_.__name__] = lambda *a: op(f, *a)
    def __neg__(s):
        return F(op(neg, s))
    def __add__(s, a):
        return F(op(add, s, a))
    def __radd__(s, a):
        return F(op(add, a, s))
    def __sub__(s, a):
        return F(op(sub, s, a))
    def __rsub__(s, a):
        return F(op(sub, a, s))
    def __mul__(s, a):
        return F(op(mul, s, a))
    def __rmul__(s, a):
        return F(op(mul, a, s))
    def __truediv__(s, a):
        return F(op(truediv, s, a))
    def __rtruediv__(s, a):
        return F(op(truediv, a, s))
    def __floordiv__(s, a):
        return F(op(floordiv, s, a))
    def __rfloordiv__(s, a):
        return F(op(floordiv, a, s))
    def __pow__(s, a):
        return F(op(pow, s, a))
    def __rpow__(s, a):
        return F(op(pow, s, a))
    def __call__(s, *a):
        return F(op(call, s, *a))
    def __getitem__(s, i):
        return F(op(getitem, s, i))
    def __setitem__(s, i, a):
        create_task(op(setitem, s, i, a))
    def __getattr__(s, i):
        return super().__getattr__(i) if i[0]=='_' else F(op(getattr, s, i))
    def __setattr__(s, i, a):
        if i[0] == '_':
            super().__setattr__(i, a)
        else:
            create_task(op(setattr, s, i, a))
```
