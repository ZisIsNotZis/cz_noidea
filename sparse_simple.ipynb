```python
from numba import njit
import numpy as np
@njit('u4(u4[:],u4)')
def index(k, q):
    for i in range(len(k)):
        if k[i] == q:
            return i
    return -1
class swarray:
    def __init__(S, s, t):
        S.s = s
        S.a = np.empty(256, f'I,{t}')
        S.i = 0
    def __getitem__(s, i):
        i = index(s.a['f0'][:s.i], i)
        return 0 if i==4294967295 else s.a['f1'][i]
    def __setitem__(s, i, v):
        j = index(s.a['f0'][:s.i], i)
        if j == 4294967295:
            if s.i == len(s.a):
                s.a.resize(len(s.a)*2, refcheck=0)
            s.a[s.i] = i, v
            s.i += 1
        else:
            s.a['f1'][j] = v
    def __delitem__(s, i):
        s.i -= 1
        s.a[index(s.a['f0'][:s.i],i)] = s.a[s.i]
```

```python
import numpy as np
class srarray:
    def __init__(S, s, t):
        S.s = s
        S.a = np.empty(256, f'I,{t}')
        S.i = 0
    def __getitem__(s, i):
        j = np.searchsorted(s.a['f0'][:s.i], i)
        return s.a['f1'][j] if s.a['f0'][j]==i else 0
    def __setitem__(s, i, v):
        j = np.searchsorted(s.a['f0'][:s.i], i)
        if s.a['f0'][j] == i:
            s.a['f1'][j] = v
        else:
            if s.i == len(s.a):
                s.a.resize(len(s.a)*2, refcheck=0)
            s.a[j+1:s.i+1] = s.a[j:s.i]
            s.a[j] = i,v
            s.i += 1
    def __delitem__(s, i):
        j = np.searchsorted(s.a['f0'][:s.i], i)
        s.a[j:s.i-1] = s.a[j+1:s.i]
        s.i -= 1
```

```python
from collections import Counter
class vec(tuple):
    __slots__ = ()
    def __add__(s, a):
        return (s[0]+a[0],s[1]+a[1]) if isinstance(a,tuple) else s
    __radd__ = __add__
class sharray(Counter):
    __slots__ = ()
    def max(s, *a):
        o = sharray()
        for i,s in s.items():
            i = tuple(i[a]for a in a)
            o[i] = max(o[i], s)
        return o
    def std(s, n, *a):
        o = sharray()
        for i,s in s.items():
            o[tuple(i[a]for a in a)] += vec((s,s*s))
        n *= n
        for i,(s,t) in o.items():
            o[i] = (s*s-t)/n
        return o
    def sum(s, *a):
        o = sharray()
        for i,s in s.items():
            o[tuple(i[a]for a in a)] += s
        return o
```
