```python
from asyncio import open_connection
_,o = await open_connection('localhost', 9999)
a = bytes(262144)
for _ in range(99999):
    o.write(a)
    await o.drain()
o.close()
```

```python
from asyncio import create_task, sleep
from collections import Counter
n = 0
c = Counter()
async def _():
    global n
    while 1:
        await sleep(5)
        print(n//5242880, 'MB/s', c)
        n = 0
create_task(_())
```

```python
from pynng import Pair0
a = bytes(67108864)
p = Pair0()
p.dial('ipc:///tmp/a')
while 1:
    await p.asend(a)
    n += len(a)
    c[len(a)] += 1
```
