```python
def filterPrime(b):
    a = next(b)
    yield a
    yield from filterPrime(b for b in b if b%a>0)
```

```python
def filterPrime(b):
    while 1:
        a = next(b)
        yield a
        b = (b for b in b if b%a>0)
```

```python
def filterPrime(b):
    a = [next(b)]
    for b in b:
        for i in a:
            if not b % i:
                break
        else:
            yield b
            a.append(b)
```

```python
import numpy as np
def filterPrime(b):
    a = np.empty(9999999, 'I')
    i = 0
    for b in b:
        if (b % a[:i]).all():
            a[i] = b
            i += 1
    return a[:i]
```

```python
%time filterPrime(np.arange(3,9999999,2,'I').reshape(-1,3)[:,1:].flat)
```

```python
def f(a, n):
    a = a.ravel()
    return a[:len(a)-len(a)%n].reshape(-1,n)
def g(a, n):
    return (a%n).astype('?').view('B')
```
