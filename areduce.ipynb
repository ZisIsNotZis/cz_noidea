```python
from asyncio import wait
async def areduce_wait(a, f):
    p = set()
    while p or len(a)>1:
        p.update(f(*a) for a in zip(a[::2], a[1::2]))
        d,p = await wait(p, return_when='FIRST_COMPLETED')
        d = [await i for i in d]
        if len(a) % 2:
            d.append(a[-1])
        a = d
        print(end='.')
    return a[0]
async def add(a, b):
    return a + b
```

```python
from datetime import datetime
t = datetime.now()
await areduce_wait(range(1048576), add)
datetime.now() - t
```

```python
async def areduce(a, f):
    q = None
    def put(a):
        nonlocal q
        if q is None:
            q = a
        else:
            @go(q, a)
            async def _(*a):
                put(await f(*a))
            q = None
    for a in a:
        put(a)
    await go.ne()
    return q
```

```python
async def add(a, b):
    return a + b
await areduce(range(100), add)
```

```python
from threading import Thread
def run(f):
    f = Thread(target=f)
    f.start()
    return f
```
