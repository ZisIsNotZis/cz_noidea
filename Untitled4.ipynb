```python
from asyncio import start_server
n = 0
async def _(i, _):
    global n
    f = open('/dev/null', 'wb')
    t = await i.read(2621440)
    while t:
        f.write(t)
        n += len(t)
        t = await i.read(2621440)
await start_server(_, 'localhost', 9999)
while 1:
    await sleep(1)
    print(n//1048576, end=' MB/s\r')
    n = 0
```

```python
from asyncio import start_unix_server, sleep
from collections import Counter
n = 0
async def _(i, _):
    global n
    c = Counter()
    t = len(await i.read(67108864))
    while t:
        n += t
        c[t] += 1
        t = len(await i.read(67108864))
    print(c)
await start_unix_server(_, '/tmp/b')
while 1:
    await sleep(1)
    print(n//1048576, end=' MB/s\r')
    n = 0
```

```python
from asyncio import create_task, sleep
n = 0
async def _():
    global n
    while 1:
        await sleep(5)
        print(n//5242880, end=' MB/s\r')
        n = 0
create_task(_())
```
